### Regression Function


The regression function $r(x)$ defines the conditional expectation of $Y$ given $X=x$. In simpler terms, it's the average value of $Y$ when $X$ is fixed at a particular value $x$. The integral form expresses this mathematically

$r(x)  = E[Y |X=x] = \int (y \cdot f(y|x)) dy$, where:
*   $y$ is a possible value of the dependent variable $Y$.
*   $f(y|x)$ is the conditional probability density function of $Y$ given $X=x$.

This function is fundamental in regression analysis, as it represents the true relationship between the predictor $X$ and the response $Y$ that statistical models aim to estimate.

#### Simple Linear Regression Model

Following the general regression function $r(x) = E[Y | X=x]$, a **Simple Linear Regression** model assumes a linear relationship between the dependent variable $Y$ and a single independent variable $X$. The model is typically expressed as:

$Y_i = \beta_0 + \beta_1 X_i + \epsilon_i$, where:
*   $Y_i$: The $i$-th observed value of the dependent (response) variable.
*   $X_i$: The $i$-th observed value of the independent (predictor) variable.
*   $\beta_0$: The **intercept**, representing the expected value of $Y$ when $X=0$.
*   $\beta_1$: The **slope**, representing the expected change in $Y$ for a one-unit change in $X$.
*   $\epsilon_i$: The **error term** for the $i$-th observation, representing random variability or noise not explained by the linear relationship. It is assumed to have a mean of zero and constant variance, and typically follows a normal distribution.

In terms of the regression function, the expected value of $Y$ given $X=x$ for a simple linear regression is:

$E[Y_i | X_i=x] = \beta_0 + \beta_1 x$

This shows how the regression function $r(x)$ takes on a specific linear form in this model.

#### Linear Regression Model

When we estimate a Linear Regression model from data, we obtain a **fitted line** which is an estimate of the true regression function $r(x)$. This fitted line is represented as:

$\hat{r}(x) = \hat{\beta}_0 + \hat{\beta}_1 x$, where:
*   $\hat{r}(x)$ is the estimated regression function, $\hat{\beta}_0$ is the estimated intercept, $\hat{\beta}_1$ is the estimated slope.

For each observed data point $(X_i, Y_i)$, we can calculate the **predicted value** (or **fitted value**) of $Y_i$, denoted as $\hat{Y}_i$, by plugging $X_i$ into the fitted line equation: $\hat{Y}_i = \hat{\beta}_0 + \hat{\beta}_1 X_i$.

The **residuals**, denoted as $e_i$, are the differences between the observed values $Y_i$ and the predicted values $\hat{Y}_i$ $e_i = Y_i - \hat{Y}_i$.

Residuals represent the part of the dependent variable that the model does not explain. In a well-fitting model, residuals should ideally be small, randomly distributed around zero, and show no systematic patterns, reflecting the assumptions about the error term $\epsilon_i$.

#### Least Squares Estimates

In simple linear regression, the goal is to find the values of $\hat{\beta}_0$ (intercept) and $\hat{\beta}_1$ (slope) that minimize the **sum of squared residuals (SSR)**. This method is known as the Ordinary Least Squares (OLS) method. The SSR is given by:

$SSR = \sum_{i=1}^{n} e_i^2 = \sum_{i=1}^{n} (Y_i - \hat{Y}_i)^2 = \sum_{i=1}^{n} (Y_i - (\hat{\beta}_0 + \hat{\beta}_1 X_i))^2$

To find the values of $\hat{\beta}_0$ and $\hat{\beta}_1$ that minimize this sum, we take partial derivatives with respect to each parameter, set them to zero, and solve the resulting system of equations:

$\frac{\partial SSR}{\partial \hat{\beta}_0} = -2 \sum_{i=1}^{n} (Y_i - \hat{\beta}_0 - \hat{\beta}_1 X_i) = 0$

$\frac{\partial SSR}{\partial \hat{\beta}_1} = -2 \sum_{i=1}^{n} X_i (Y_i - \hat{\beta}_0 - \hat{\beta}_1 X_i) = 0$

Solving these normal equations yields the following **Least Squares Estimates**:

**Estimate for the Slope ($\hat{\beta}_1$)**:
$\hat{\beta}_1 = \frac{\sum_{i=1}^{n} (X_i - \bar{X})(Y_i - \bar{Y})}{\sum_{i=1}^{n} (X_i - \bar{X})^2} = \frac{S_{XY}}{S_{XX}}$ where:
    
$\bar{X}$ is the mean of $X_i$. $\bar{Y}$ is the mean of $Y_i$. $S_{XY}$ is the sum of cross-products of deviations. $S_{XX}$ is the sum of squares of deviations for $X$.


**Estimate for the Intercept ($\hat{\beta}_0$)**: $\hat{\beta}_0 = \bar{Y} - \hat{\beta}_1 \bar{X}$

These formulas provide the unique best linear unbiased estimates (BLUE) of the true population parameters $\beta_0$ and $\beta_1$ under the Gauss-Markov assumptions.

#### Maximum Likelihood Estimates (MLE)

MLE is another powerful method for estimating model parameters. It seeks to find the parameter values that maximize the likelihood function, i.e., the probability of observing the given data under the assumed statistical model.

For the simple linear regression model $Y_i = \beta_0 + \beta_1 X_i + \epsilon_i$, if we make the crucial assumption that the error terms $\epsilon_i$ are iid. normal rvs with mean 0 and constant variance $\sigma^2$, i.e., $\epsilon_i \sim N(0, \sigma^2)$, then $Y_i$ itself is normally distributed with mean $\beta_0 + \beta_1 X_i$ and variance $\sigma^2$: $Y_i \sim N(\beta_0 + \beta_1 X_i, \sigma^2)$.

The likelihood function for the observed data $(X_1, Y_1), \dots, (X_n, Y_n)$ is the product of the probability density functions of each $Y_i$:

$L(\beta_0, \beta_1, \sigma^2 | X, Y) = \prod_{i=1}^{n} \frac{1}{\sqrt{2\pi\sigma^2}} e^{-\frac{(Y_i - (\beta_0 + \beta_1 X_i))^2}{2\sigma^2}}$

To find the MLEs, we typically maximize the natural logarithm of the likelihood function (the log-likelihood), as it simplifies calculations and its maximum occurs at the same parameter values as the original likelihood function:

$\ln L = -\frac{n}{2} \ln(2\pi) - \frac{n}{2} \ln(\sigma^2) - \frac{1}{2\sigma^2} \sum_{i=1}^{n} (Y_i - (\beta_0 + \beta_1 X_i))^2$

To find the values of $\beta_0$ and $\beta_1$ that maximize $\ln L$, we need to minimize the term $\sum_{i=1}^{n} (Y_i - (\beta_0 + \beta_1 X_i))^2$. This term is precisely the Sum of Squared Residuals (SSR) that the Ordinary Least Squares (OLS) method minimizes.

Under the assumption of i.i.d. normally distributed error terms, the MLE for the regression coefficients ($\hat{\beta}_0$ and $\hat{\beta}_1$) are identical to the OLS estimates. MLE also provides an estimate for the error variance $\sigma^2$.

#### Properties of the Least Squares Estimates (LSE)

The LSE, $\hat{\beta}_0$ and $\hat{\beta}_1$, have several desirable statistical properties under a specific set of assumptions known as the **Gauss-Markov Assumptions**. These assumptions ensure that OLS is a powerful and reliable estimation method. When these assumptions hold, the LSE are considered the **Best Linear Unbiased Estimators (BLUE)**.

Here are the key properties:

1.  **Linearity**: The estimators $\hat{\beta}_0$ and $\hat{\beta}_1$ are linear functions of the observed response variable $Y$.

2.  **Unbiasedness**: $E[\hat{\beta}_0] = \beta_0$ and $E[\hat{\beta}_1] = \beta_1$. This means that, on average, the LSE will equal the true population parameters. In other words, if we were to take many samples and compute the estimates for each, the average of these estimates would converge to the true parameter values.

    *Assumptions for Unbiasedness*: Linearity in parameters, random sampling, $E[\epsilon_i | X_i] = 0$ (zero conditional mean of errors), and sufficient variation in $X$ (not all $X_i$ are the same).

3.  **Efficiency** (BLUE): Among all linear and unbiased estimators, the LSE have the smallest variance. This means that the LSE provide the most precise estimates (least spread out) compared to any other estimator that is both linear and unbiased.
    
    *Additional Assumption for Efficiency (relative to other linear unbiased estimators)*: $Var(\epsilon_i | X_i) = \sigma^2$ (homoscedasticity - constant variance of errors).

4.  **Consistency**: As the sample size ($n$) approaches infinity, the LSE converge to the true population parameters. This means that with a sufficiently large amount of data, the estimates will be very close to the true values.
    
    *Assumption for Consistency*: Requires largely the same assumptions as unbiasedness, plus some regularity conditions regarding the distribution of $X$.

5.  **Normality (under Normal Errors)**: If the error terms $\epsilon_i$ are normally distributed (i.i.d. $N(0, \sigma^2)$), then the LSE $\hat{\beta}_0$ and $\hat{\beta}_1$ are also normally distributed. This property is crucial for hypothesis testing and constructing confidence intervals for the regression coefficients. Even without normally distributed errors, the CLT often implies that LSE are approximately normally distributed for large sample sizes.

#### Prediction and Prediction Interval

Once a linear regression model has been estimated, it can be used to make predictions for new observations. There are two main types of predictions in this context: point predictions and prediction intervals.

**1. Point Prediction ($\hat{Y}_{new}$)**

A **point prediction** is a single value that represents the best estimate of the dependent variable ($Y$) for a new, unseen value of the independent variable ($X_{new}$). It is calculated by plugging $X_{new}$ into the fitted regression equation: $\hat{Y}_{new} = \hat{\beta}_0 + \hat{\beta}_1 X_{new}$

This point estimate represents the expected value of $Y$ for a given $X_{new}$, based on the estimated linear relationship.

**2. Prediction Interval (PI)**

A prediction interval (PI) is a range that is likely to contain a single, new observation of the dependent variable $Y$ for a given value of $X_{new}$. Unlike a confidence interval for the *mean* response, a prediction interval accounts for both the uncertainty in estimating the regression line (similar to a confidence interval for the mean) AND the inherent variability of individual observations around the regression line (the error term $\epsilon$).

For a given confidence level (e.g., 95%), a prediction interval for a new observation $\hat{Y}_{new}$ at $X_{new}$ is typically calculated as:

$\hat{Y}_{new} \pm t_{\alpha/2, n-2} \cdot SE_{pred}$
where:
*   $\hat{Y}_{new}$ is the point prediction.
*   $t_{\alpha/2, n-2}$ is the critical t-value for the desired confidence level and degrees of freedom ($n-2$ for simple linear regression).
*   $SE_{pred}$ is the standard error of the prediction, which reflects the variability of a single new observation. It is larger than the standard error of the mean response because it includes the error variance ($s^2$) of individual observations.

The formula for $SE_{pred}$ is:

$SE_{pred} = \sqrt{s^2 \left(1 + \frac{1}{n} + \frac{(X_{new} - \bar{X})^2}{\sum (X_i - \bar{X})^2}\right)}$

Here, $s^2$ is the estimated variance of the error terms.

Key Differences from Confidence Interval for the Mean Response:

*   **Prediction Interval (PI)**: Estimates a range for an *individual new observation*. It is generally wider because it accounts for the individual error term $\epsilon$.
*   **Confidence Interval (CI) for the Mean Response**: Estimates a range for the *average* value of $Y$ at a given $X$. It is narrower because it only accounts for the uncertainty in the estimate of the mean, not the variability of individual observations around that mean.

Prediction intervals are crucial for understanding the potential range of outcomes for future observations, providing a more complete picture of uncertainty than point predictions alone.

#### Multiple Linear Regression Model

**Multiple Linear Regression** is an extension of simple linear regression that models the relationship between a dependent variable ($Y$) and *two or more* independent (or predictor) variables ($X_1, X_2, \dots, X_p$). It is still considered a linear model because the relationship is linear in the parameters ($\beta$ coefficients).

The general form of a multiple linear regression model is:

$Y_i = \beta_0 + \beta_1 X_{i1} + \beta_2 X_{i2} + \dots + \beta_p X_{ip} + \epsilon_i$

Where:
*   $Y_i$: The $i$-th observed value of the dependent (response) variable.
*   $X_{i1}, X_{i2}, \dots, X_{ip}$: The $i$-th observed values of the $p$ independent variables.
*   $\beta_0$: The **intercept**, representing the expected value of $Y$ when all independent variables are zero.
*   $\beta_j$: The **partial slope coefficient** for the $j$-th independent variable ($X_j$). It represents the expected change in $Y$ for a one-unit change in $X_j$, *holding all other independent variables constant*.
*   $\epsilon_i$: The **error term** for the $i$-th observation, representing random variability not explained by the linear relationship. Similar to simple linear regression, it is assumed to have a mean of zero, constant variance, and typically follows a normal distribution.

For example, if we have three independent variables, say `House_Size` ($X_1$), `Num_Bedrooms` ($X_2$), and `Distance_to_City` ($X_3$), the model would be:

$Y_i = \beta_0 + \beta_1 \text{House_Size}_i + \beta_2 \text{Num_Bedrooms}_i + \beta_3 \text{Distance_to_City}_i + \epsilon_i$

In terms of the regression function, the expected value of $Y$ given specific values for the independent variables is:

$E[Y_i | X_{i1}=x_1, X_{i2}=x_2, \dots, X_{ip}=x_p] = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \dots + \beta_p x_p$

Multiple linear regression allows us to analyze how multiple factors simultaneously influence the dependent variable, providing a more comprehensive understanding of complex relationships in data.

#### Model Selection

**Model selection** is the process of choosing the best statistical model from a set of candidate models. In the context of linear regression, this often involves deciding which predictor variables to include in the model, or choosing between different forms of a model (e.g., simple vs. multiple linear regression, or even considering polynomial terms).

The goal of model selection is to find a model that accurately captures the underlying relationships in the data while avoiding overfitting (a model that is too complex and fits the noise in the training data) and underfitting (a model that is too simple and fails to capture important patterns).

Several criteria and techniques are used for model selection, each with its strengths and weaknesses:

1.  **R-squared ($R^2$) and Adjusted R-squared (Adj. $R^2$)**:
    *   **$R^2$**: Measures the proportion of the variance in the dependent variable that is predictable from the independent variables. A higher $R^2$ generally indicates a better fit. However, $R^2$ always increases with the addition of more variables, even if they are not significant.
    *   **Adjusted $R^2$**: A modified version of $R^2$ that accounts for the number of predictors in the model. It increases only if the new term improves the model more than would be expected by chance, making it more suitable for comparing models with different numbers of predictors.

2.  **Akaike Information Criterion (AIC)**:
    *   A measure of the relative quality of statistical models for a given set of data. AIC estimates the relative amount of information lost by a given model. The model with the lowest AIC is generally preferred.
    *   $AIC = 2k - 2\ln(L)$, where $k$ is the number of parameters in the model and $L$ is the maximum value of the likelihood function for the model.

3.  **Bayesian Information Criterion (BIC)**:
    *   Similar to AIC, BIC is also used for model selection. It penalizes models with more parameters more heavily than AIC, especially for larger datasets. The model with the lowest BIC is preferred.
    *   $BIC = k\ln(n) - 2\ln(L)$, where $k$ is the number of parameters, $n$ is the number of observations, and $L$ is the maximum likelihood.

4.  **Cross-Validation**:
    *   A robust technique for evaluating how well a model generalizes to new data. It involves splitting the dataset into multiple subsets, training the model on some subsets, and testing it on others. Common methods include K-fold cross-validation and Leave-One-Out Cross-Validation (LOOCV).

5.  **Hypothesis Testing (e.g., F-test, t-test)**:
    *   Used to test the significance of individual predictors or groups of predictors. For example, an F-test can be used to compare the overall fit of two nested models (where one model is a subset of the other), and t-tests are used to assess the significance of individual regression coefficients.

Choosing the right model is an iterative process that involves balancing model complexity with explanatory power and predictive accuracy.

#### Prediction Risk and Training Error

When we build a statistical model, like a linear regression model, our main goal is often to make accurate predictions on new, unseen data. To understand how well our model performs, we use concepts like **prediction risk** and **training error**.

**1. Training Error**

The **training error** (also known as *in-sample error* or *apparent error*) is a measure of how well our model fits the data it was trained on. It's calculated by comparing the model's predictions to the actual values in the training dataset. For a regression model, a common measure of training error is the **Mean Squared Error (MSE)** on the training data:

$MSE_{train} = \frac{1}{n} \sum_{i=1}^{n} (Y_i - \hat{Y}_i)^2$

Where $Y_i$ are the actual values and $\hat{Y}_i$ are the predictions made by the model on the training data.

A low training error suggests that the model fits the observed data well. However, it doesn't necessarily mean the model will perform well on new data.

**2. Prediction Risk (or Generalization Error)**

The **prediction risk** (also known as *generalization error* or *test error*) is the expected error of the model when applied to new, unseen data. It's a more realistic measure of a model's true performance, as it reflects how well the model generalizes beyond the specific dataset it was trained on.

Minimizing prediction risk is the ultimate goal in most predictive modeling tasks. The true prediction risk cannot be directly calculated because we don't have access to all possible future data. Instead, we estimate it using techniques like cross-validation or by holding out a separate test set.

**The Relationship: Training Error vs. Prediction Risk**

It's common for a model to have a very low training error but a high prediction risk. This situation is known as **overfitting**. Overfitting occurs when a model learns the training data too well, capturing not only the underlying patterns but also the noise and random fluctuations specific to that dataset.

When a model is overfit:
*   It performs exceptionally well on the training data.
*   Its performance drops significantly on new, unseen data.

Conversely, a model with a high training error might also have a high prediction risk, which is often an indication of **underfitting**. Underfitting occurs when a model is too simple to capture the underlying relationships in the data.

The key challenge in model building is to find a balance between minimizing training error and maintaining a low prediction risk, which often involves techniques discussed in model selection (like using adjusted R-squared, AIC, BIC, or cross-validation) to choose a model that generalizes well.

##### LASSO Regression (Least Absolute Shrinkage and Selection Operator)

LASSO Regression is a type of linear regression that uses shrinkage. Shrinkage methods are regularized techniques that aim to improve the prediction accuracy and interpretability of the statistical model by shrinking some of the regression coefficients towards zero. Specifically, LASSO performs both variable selection and regularization to enhance the prediction accuracy and interpretability of the model.

Why LASSO?
1.  **Feature Selection**: LASSO drives the coefficients of less important features exactly to zero, effectively performing automatic feature selection. This means that variables with zero coefficients are excluded from the model.
2.  **Regularization**: It helps in preventing overfitting by adding a penalty term to the cost function, thereby shrinking the coefficients. This is especially useful when you have a large number of features, some of which might be highly correlated or irrelevant.
3.  **Improved Model Interpretability**: By setting coefficients of irrelevant features to zero, LASSO creates simpler, more interpretable models.

Mathematical Formulation

The standard Ordinary Least Squares (OLS) method minimizes the sum of squared residuals:

$SSR = \sum_{i=1}^{n} (Y_i - (\beta_0 + \sum_{j=1}^{p} \beta_j X_{ij}))^2$

LASSO regression adds a penalty term to this objective function. The penalty is proportional to the absolute value of the magnitude of the regression coefficients ($\|\beta_j\|$). The objective function that LASSO minimizes is:

$J(\beta) = \sum_{i=1}^{n} (Y_i - (\beta_0 + \sum_{j=1}^{p} \beta_j X_{ij}))^2 + \lambda \sum_{j=1}^{p} |\beta_j|$

Where:
*   $\sum_{i=1}^{n} (Y_i - (\beta_0 + \sum_{j=1}^{p} \beta_j X_{ij}))^2$ is the **Sum of Squared Residuals (SSR)**, representing the model's fit to the data.
*   $\lambda \sum_{j=1}^{p} |\beta_j|$ is the **LASSO penalty term** (also known as the L1 penalty).
*   $\lambda$ (lambda) is the **tuning parameter** (or regularization parameter). It controls the strength of the penalty.
    *   If $\lambda = 0$, the LASSO objective function becomes equivalent to OLS, and no shrinkage occurs.
    *   As $\lambda$ increases, more coefficients are shrunk towards zero, and eventually, some become exactly zero, leading to feature selection.

Key Characteristics
*   **L1 Regularization**: The penalty term uses the L1 norm of the coefficients ($\|\beta\|_1 = \sum |\beta_j|$), which is why LASSO is also called L1 regularization.
*   **Sparsity**: The L1 penalty has a tendency to produce sparse models, meaning many coefficients become exactly zero, leading to simpler models with fewer features.

Choosing an optimal value for $\lambda$ is critical and is typically done through techniques like cross-validation.

#### Practical Example: Wald Test for $\beta_1 = m \cdot \beta_0$

Let's put the Wald test into practice. We'll generate some synthetic data, fit a simple linear regression model, and then test the hypothesis $H_0: \beta_1 = m \cdot \beta_0$ vs $H_1: \beta_1 \neq m \cdot \beta_0$ for a given value of $m$.

#### Constructing a Wald Test for $H_0: \beta_1 = m \cdot \beta_0$

To construct a Wald test for the hypothesis $H_0: \beta_1 = m \cdot \beta_0$ against the alternative $H_1: \beta_1 \neq m \cdot \beta_0$, we follow these steps:

1.  **Define the Hypothesis**: The hypothesis can be re-arranged to define a new parameter, $\theta$, as $H_0: \beta_1 - m \beta_0 = 0 \quad \Rightarrow \quad H_0: \theta = 0$
where $\theta = \beta_1 - m \beta_0$.

2.  **Estimate the Parameter of Interest**: Using the estimated coefficients from a simple linear regression model, $\hat{\beta}_0$ and $\hat{\beta}_1$, the estimate for $\theta$ is: $\hat{\theta} = \hat{\beta}_1 - m \hat{\beta}_0$.

3.  **Calculate the Variance of the Estimator**: To find the standard error of $\hat{\theta}$, we first need its variance. Using the properties of variances, specifically $Var(aX + bY) = a^2Var(X) + b^2Var(Y) + 2abCov(X, Y)$, we can derive $Var(\hat{\theta})$:
$Var(\hat{\theta}) = Var(\hat{\beta}_1 - m \hat{\beta}_0) = Var(\hat{\beta}_1) + (-m)^2 Var(\hat{\beta}_0) + 2(-m)Cov(\hat{\beta}_1, \hat{\beta}_0)$
$Var(\hat{\theta}) = Var(\hat{\beta}_1) + m^2 Var(\hat{\beta}_0) - 2m Cov(\hat{\beta}_1, \hat{\beta}_0)$.
The individual variances $Var(\hat{\beta}_0)$ and $Var(\hat{\beta}_1)$, and their covariance $Cov(\hat{\beta}_1, \hat{\beta}_0)$, are obtained from the estimated covariance matrix of the regression coefficients (e.g., results.cov_params() in statsmodels). Let's denote them as $\hat{Var}(\hat{\beta}_0)$, $\hat{Var}(\hat{\beta}_1)$, and $\hat{Cov}(\hat{\beta}_1, \hat{\beta}_0)$.

4.  **Calculate the Standard Error of the Estimator**: The standard error of $\hat{\theta}$ is the square root of its estimated variance:

    $SE(\hat{\theta}) = \sqrt{\hat{Var}(\hat{\theta})} = \sqrt{\hat{Var}(\hat{\beta}_1) + m^2 \hat{Var}(\hat{\beta}_0) - 2m \hat{Cov}(\hat{\beta}_1, \hat{\beta}_0)}$

5.  **Compute the Wald Test Statistic**: Under the null hypothesis, the Wald test statistic for $\theta = 0$ follows a t-distribution (for small samples, assuming normal errors) or a Z-distribution (for large samples):

    $W = \frac{\hat{\theta} - 0}{SE(\hat{\theta})} = \frac{\hat{\beta}_1 - m \hat{\beta}_0}{SE(\hat{\beta}_1 - m \hat{\beta}_0)}$

    The degrees of freedom for this t-distribution are typically $n - k - 1$, where $n$ is the number of observations and $k$ is the number of predictors (1 for simple linear regression, so $n-2$).

6.  **Determine the P-value**: For a two-sided test ($H_1: \theta \neq 0$), the p-value is calculated as:

    $p = 2 \cdot P(T > |W|)$
    where $T$ is a random variable following a t-distribution with the appropriate degrees of freedom.

7.  **Make a Decision**: Compare the p-value to a predetermined significance level ($\alpha$). If $p < \alpha$, reject the null hypothesis. Otherwise, fail to reject the null hypothesis.


```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

# 1. Generate Sample Data
np.random.seed(42)
X = np.random.rand(100) * 10 # Independent variable
# True relationship: Y = 2 + 0.5 * X + noise
# Let's say we want to test H0: beta_1 = 0.2 * beta_0, so m = 0.2
beta_0_true = 2
beta_1_true = 0.5
Y = beta_0_true + beta_1_true * X + np.random.randn(100) * 1.5

# Add a constant to X for the intercept term in statsmodels
X_const = sm.add_constant(X)

df_data = pd.DataFrame({'X': X, 'Y': Y, 'const': 1})
display(df_data.head())
```

Fit a simple linear regression model using statsmodels.

```python
# 2. Fit a Simple Linear Regression Model
model = sm.OLS(Y, X_const)
results = model.fit()

# Extract estimated coefficients
beta_0_hat = results.params[0] # Intercept
beta_1_hat = results.params[1] # Slope
```

Construct the Wald test for the hypothesis $H_0: \beta_1 = m \cdot \beta_0$. For this example, let's choose $m=0.2$ as stated in the comment when generating the data.

```python
# 3. Implement the Wald Test
m = 0.2 # The value for 'm' in H0: Beta_1 = m * Beta_0

# Estimated linear combination (theta_hat = beta_1_hat - m * beta_0_hat)
theta_hat = beta_1_hat - m * beta_0_hat

# Get the covariance matrix of the coefficients
cov_matrix = results.cov_params()

# Extract variances and covariance using integer indexing for numpy array
# 'const' is at index 0, 'x1' is at index 1
var_beta_0_hat = cov_matrix[0, 0]
var_beta_1_hat = cov_matrix[1, 1]
cov_beta_0_beta_1_hat = cov_matrix[0, 1] # or cov_matrix[1, 0]

# Calculate the variance of theta_hat
# Var(theta_hat) = Var(beta_1_hat) + m^2 * Var(beta_0_hat) - 2m * Cov(beta_1_hat, beta_0_hat)
var_theta_hat = var_beta_1_hat + (m**2 * var_beta_0_hat) - (2 * m * cov_beta_0_beta_1_hat)

# Standard error of theta_hat
se_theta_hat = np.sqrt(var_theta_hat)

# Wald test statistic
wald_statistic = theta_hat / se_theta_hat

# Degrees of freedom
df = results.df_resid # n - p - 1, where p is number of predictors (1 for simple LR)

# P-value (two-sided test)
from scipy.stats import t
p_value = 2 * t.cdf(-abs(wald_statistic), df)

# Decision based on a significance level (e.g., alpha = 0.05)
alpha = 0.05
if p_value < alpha:
    print(f"\nAt alpha = {alpha}, we reject the null hypothesis.")
else:
    print(f"\nAt alpha = {alpha}, we fail to reject the null hypothesis.")
```

#### Model Selection using Mallow's $C_p$ Criterion

Mallow's $C_p$ is a criterion used for model selection in linear regression. It helps in finding a subset of predictors that are adequate for prediction, aiming to strike a balance between the goodness of fit and the complexity of the model. A model with a $C_p$ value close to the number of parameters ($p$) (including the intercept) indicates a good balance.

The formula for Mallow's $C_p$ is:

$C_p = \frac{SSR_p}{\hat{\sigma}^2} - (n - 2p)$, where:
*   $SSR_p$: Sum of squared residuals for the sub-model with $p$ parameters.
*   $\hat{\sigma}^2$: MSE of the full model (with all predictors), used as an estimate of the true error variance.
*   $n$: Number of observations.
*   $p$: Number of parameters in the sub-model (including the intercept).

We implement functions to calculate $C_p$ and then apply it to select the best sub-model using various search strategies.

```python
import itertools

# Estimate sigma_squared from the full model's MSE (multi_results.scale is MSE of residuals)
# For Mallow's Cp, we use the MSE of the full model as the estimate for sigma^2
sigma_squared_full = multi_results.scale

def calculate_cp(model_results, sigma_squared_full, n):
    """Calculates Mallow's Cp for a given statsmodels OLS results object."""
    ssr_p = model_results.ssr # Sum of squared residuals for the sub-model
    p = len(model_results.params) # Number of parameters in the sub-model (includes intercept)
    cp = (ssr_p / sigma_squared_full) - (n - 2 * p)
    return cp

# Get the names of the potential independent variables (excluding 'const')
all_features = ['House_Size', 'Num_Bedrooms', 'Distance_to_City']
n = len(df_multi)

# Store results for all possible models
cp_results = []

print("\n--- Mallow's Cp for All Possible Sub-models ---")

# Iterate through all possible combinations of features
for i in range(len(all_features) + 1):
    for subset in itertools.combinations(all_features, i):
        current_features = list(subset)

        if not current_features:
            # Intercept-only model
            X_subset = sm.add_constant(pd.DataFrame({'const': np.ones(n)}))
            model_name = "Intercept Only"
        else:
            X_subset = sm.add_constant(df_multi[current_features])
            model_name = " + ".join(current_features)

        # Fit the model
        temp_model = sm.OLS(Y_multi, X_subset)
        temp_results = temp_model.fit()

        # Calculate Cp
        cp_value = calculate_cp(temp_results, sigma_squared_full, n)
        bic_value = temp_results.bic # Also store BIC for later comparison

        cp_results.append({
            'Features': model_name,
            'Num_Params': len(temp_results.params),
            'Cp': cp_value,
            'BIC': bic_value,
            'R_squared_adj': temp_results.rsquared_adj,
            'Model': temp_results # Store model for summary if needed
        })

# Convert to DataFrame for easy viewing and sorting
cp_df = pd.DataFrame(cp_results)

# --- Mallow's Cp Analysis ---
cp_df_sorted_by_cp = cp_df.sort_values(by='Cp')
print("Sorted by Mallow's Cp:")
display(cp_df_sorted_by_cp[['Features', 'Num_Params', 'Cp', 'R_squared_adj']].round(3))

# Identify the best model based on Cp
best_cp_model_row = cp_df_sorted_by_cp.loc[cp_df_sorted_by_cp['Cp'].idxmin()]
print(f"\nBest model by Mallow's Cp: {best_cp_model_row['Features']} (Cp: {best_cp_model_row['Cp']:.3f})")
print("Summary of Best Cp Model:")

# Determine X for the best Cp model
if best_cp_model_row['Features'] == "Intercept Only":
    X_best_cp = sm.add_constant(pd.DataFrame({'const': np.ones(n)}))
elif best_cp_model_row['Features'] == "House_Size + Num_Bedrooms + Distance_to_City":
    # It's the full model, use the already fitted multi_results
    X_best_cp = X_multi
else:
    X_best_cp = sm.add_constant(df_multi[best_cp_model_row['Features'].split(' + ')])

final_cp_model = sm.OLS(Y_multi, X_best_cp).fit()
print(final_cp_model.summary())

# --- Bayesian Information Criterion (BIC) Analysis ---
print("\n--- Bayesian Information Criterion (BIC) Analysis ---")

bic_df_sorted_by_bic = cp_df.sort_values(by='BIC')
print("Sorted by BIC:")
display(bic_df_sorted_by_bic[['Features', 'Num_Params', 'BIC', 'R_squared_adj']].round(3))

# Identify the best model based on BIC
best_bic_model_row = bic_df_sorted_by_bic.loc[bic_df_sorted_by_bic['BIC'].idxmin()]
print(f"\nBest model by BIC: {best_bic_model_row['Features']} (BIC: {best_bic_model_row['BIC']:.3f})")
print("Summary of Best BIC Model:")

# Determine X for the best BIC model
if best_bic_model_row['Features'] == "Intercept Only":
    X_best_bic = sm.add_constant(pd.DataFrame({'const': np.ones(n)}))
elif best_bic_model_row['Features'] == "House_Size + Num_Bedrooms + Distance_to_City":
    X_best_bic = X_multi
else:
    X_best_bic = sm.add_constant(df_multi[best_bic_model_row['Features'].split(' + ')])

final_bic_model = sm.OLS(Y_multi, X_best_bic).fit()
print(final_bic_model.summary())

# --- Comparison of Mallow's Cp and BIC Results ---
print("\n--- Comparison: Mallow's Cp vs. BIC ---")
print(f"Best model by Mallow's Cp: {best_cp_model_row['Features']} (Cp: {best_cp_model_row['Cp']:.3f})")
print(f"Best model by BIC: {best_bic_model_row['Features']} (BIC: {best_bic_model_row['BIC']:.3f})")

if best_cp_model_row['Features'] == best_bic_model_row['Features']:
    print("\nBoth Mallow's Cp and BIC recommend the same model.")
else:
    print("\nMallow's Cp and BIC recommend different models. BIC typically favors simpler models due to its stronger penalty for the number of parameters (k*ln(n) vs. 2k for AIC/Cp).")


# --- Forward Stepwise Selection (using Cp) ---
print("\n--- Forward Stepwise Selection (using Cp) ---")

def forward_stepwise_cp(X, y, all_features, sigma_squared_full, n):
    remaining_features = set(all_features)
    selected_features = []
    best_cp = np.inf
    best_model_so_far = None
    step_models = []

    while remaining_features:
        candidates = []
        for feature in remaining_features:
            current_selection = selected_features + [feature]
            if not current_selection:
                X_current = sm.add_constant(pd.DataFrame({'const': np.ones(n)}))
            else:
                X_current = sm.add_constant(X[current_selection])
            temp_model = sm.OLS(y, X_current).fit()
            cp = calculate_cp(temp_model, sigma_squared_full, n)
            candidates.append((cp, feature, temp_model))

        candidates.sort()
        if not candidates: # No more features to add
            break

        min_cp, best_feature, current_best_model = candidates[0]

        if min_cp < best_cp:
            best_cp = min_cp
            selected_features.append(best_feature)
            remaining_features.remove(best_feature)
            best_model_so_far = current_best_model
            step_models.append({'Features': ' + '.join(selected_features), 'Cp': best_cp, 'Model': best_model_so_far})
        else:
            # Adding more features increases Cp, so stop
            break

    if not step_models:
        # Handle case where intercept-only is best and no features are added
        X_intercept = sm.add_constant(pd.DataFrame({'const': np.ones(n)}))
        intercept_model = sm.OLS(y, X_intercept).fit()
        cp_intercept = calculate_cp(intercept_model, sigma_squared_full, n)
        return cp_intercept, intercept_model, [{'Features': 'Intercept Only', 'Cp': cp_intercept, 'Model': intercept_model}]

    return best_cp, best_model_so_far, step_models

best_cp_forward, forward_model, forward_steps = forward_stepwise_cp(df_multi[all_features], Y_multi, all_features, sigma_squared_full, n)
print("Forward Stepwise Selection Steps (using Cp):")
for step in forward_steps:
    print(f"  - Features: {step['Features']}, Cp: {step['Cp']:.3f}")
print(f"Best model from Forward Stepwise (Cp): {forward_steps[-1]['Features']} (Cp: {best_cp_forward:.3f})")
print("Summary of Best Forward Stepwise Model:")
print(forward_model.summary())

# --- Backward Stepwise Selection (using Cp) ---
print("\n--- Backward Stepwise Selection (using Cp) ---")

def backward_stepwise_cp(X, y, all_features, sigma_squared_full, n):
    selected_features = list(all_features) # Start with full model
    best_cp = np.inf
    best_model_so_far = None
    step_models = []

    # Initial full model
    X_current = sm.add_constant(X[selected_features])
    full_model = sm.OLS(y, X_current).fit()
    initial_cp = calculate_cp(full_model, sigma_squared_full, n)
    best_cp = initial_cp
    best_model_so_far = full_model
    step_models.append({'Features': ' + '.join(selected_features), 'Cp': initial_cp, 'Model': full_model})

    while len(selected_features) > 0:
        candidates = []
        for feature_to_remove in selected_features:
            current_selection = [f for f in selected_features if f != feature_to_remove]

            if not current_selection:
                X_current = sm.add_constant(pd.DataFrame({'const': np.ones(n)}))
            else:
                X_current = sm.add_constant(X[current_selection])

            temp_model = sm.OLS(y, X_current).fit()
            cp = calculate_cp(temp_model, sigma_squared_full, n)
            candidates.append((cp, feature_to_remove, temp_model, current_selection))

        if not candidates: # Only intercept remains
            break

        candidates.sort()
        min_cp_after_removal, removed_feature, current_best_model, new_selection = candidates[0]

        # If removing a feature improves (decreases) Cp, or keeps it close (for tie-breaking)
        if min_cp_after_removal < best_cp or (min_cp_after_removal == best_cp and len(new_selection) < len(selected_features) - 1): # Added tie-breaking to prefer simpler models
            best_cp = min_cp_after_removal
            selected_features = new_selection
            best_model_so_far = current_best_model
            step_models.append({'Features': ' + '.join(selected_features) if selected_features else 'Intercept Only', 'Cp': best_cp, 'Model': best_model_so_far})
        else:
            # Removing the best candidate feature does not improve Cp, so stop
            break

    return best_cp, best_model_so_far, step_models

best_cp_backward, backward_model, backward_steps = backward_stepwise_cp(df_multi[all_features], Y_multi, all_features, sigma_squared_full, n)
print("Backward Stepwise Selection Steps (using Cp):")
for step in backward_steps:
    print(f"  - Features: {step['Features']}, Cp: {step['Cp']:.3f}")
print(f"Best model from Backward Stepwise (Cp): {backward_steps[-1]['Features']} (Cp: {best_cp_backward:.3f})")
print("Summary of Best Backward Stepwise Model:")
print(backward_model.summary())
```

#### LASSO Regression Implementation

Now, we will implement LASSO (Least Absolute Shrinkage and Selection Operator) Regression. LASSO is a regularization technique that performs feature selection by shrinking some coefficients to zero. This is particularly useful for building sparser models and reducing overfitting.

We will use sklearn.linear_model.Lasso for this implementation and utilize cross-validation to find the optimal regularization parameter alpha (which is equivalent to lambda in the mathematical formulation).

```python
from sklearn.linear_model import LassoCV, LassoLarsCV
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Prepare the data
X_lasso = df_multi[['House_Size', 'Num_Bedrooms', 'Distance_to_City']]
Y_lasso = df_multi['House_Price']

# It's important to scale the features for Lasso regularization
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_lasso)

# Split data into training and testing sets for evaluating generalization
X_train, X_test, Y_train, Y_test = train_test_split(X_scaled, Y_lasso, test_size=0.2, random_state=42)

# Initialize LassoCV model
# alphas: list of alphas to try. If None, default alphas are used.
# cv: number of folds in cross-validation
lasso_model = LassoCV(alphas=None, cv=10, random_state=42, max_iter=10000)

# Fit the model to the scaled training data
lasso_model.fit(X_train, Y_train)

# Best alpha found by cross-validation
best_alpha = lasso_model.alpha_
print(f"Optimal alpha (lambda) found by LassoCV: {best_alpha:.4f}")

# Coefficients for the best alpha
lasso_coefficients = dict(zip(X_lasso.columns, lasso_model.coef_))
print("\nLasso Coefficients (for optimal alpha):")
for feature, coef in lasso_coefficients.items():
    print(f"  {feature}: {coef:.4f}")

# Intercept
lasso_intercept = lasso_model.intercept_
print(f"  Intercept: {lasso_intercept:.4f}")

# Evaluate the model on the test set
r_squared_test = lasso_model.score(X_test, Y_test)
print(f"\nR-squared on test data: {r_squared_test:.4f}")

# Predict on the test set
Y_pred_lasso = lasso_model.predict(X_test)

# Plotting coefficients path (optional but informative)
# This helps visualize how coefficients change with different alpha values
plt.figure(figsize=(12, 6), dpi=60)
# Use LassoLarsCV to get the coef_path_
glmnet = LassoLarsCV(cv=10, max_iter=10000)
glmnet.fit(X_scaled, Y_lasso)


alphas = glmnet.alphas_
coefs = glmnet.coef_path_.T

ax = plt.gca()
ax.plot(alphas, coefs)
ax.set_xscale('log')
plt.xlabel('alpha (log scale)')
plt.ylabel('Coefficients')
plt.title('Lasso Coefficients Path')
plt.legend(X_lasso.columns)
plt.axvline(best_alpha, linestyle='--', color='red', label=f'Optimal alpha ({best_alpha:.4f})')
plt.grid(True)
plt.show()
```
